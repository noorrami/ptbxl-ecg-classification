models for ECG classification using the PTB-XL dataset.

```python
import numpy as np
import pandas as pd
import sklearn
import matplotlib
import mne
import wfdb
import torch

versions = {
    "NumPy": np.__version__,
    "Pandas": pd.__version__,
    "scikit-learn": sklearn.__version__,
    "Matplotlib": matplotlib.__version__,
    "MNE": mne.__version__,
    "WFDB": wfdb.__version__,
    "PyTorch": torch.__version__,
}

for name, version in versions.items():
    print(f"{name}: {version}")
```

```python
import random

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

print("Random seed configured:", SEED)
```

```python
from google.colab import drive
from pathlib import Path

drive.mount("/content/drive")

PROJECT_DRIVE = Path(
    "/content/drive/MyDrive/ML_Master_Project"
)

folders = [
    "data/raw",
    "data/processed",
    "artifacts",
    "results",
    "backups",
]

for folder in folders:
    (PROJECT_DRIVE / folder).mkdir(
        parents=True,
        exist_ok=True
    )

print("Project storage ready:", PROJECT_DRIVE)
```

```python
import json
from datetime import datetime, timezone

environment_info = {
    "project": "PTB-XL ECG Classification",
    "python_version": platform.python_version(),
    "torch_version": torch.__version__,
    "device": str(device),
    "seed": SEED,
    "created_at_utc": datetime.now(
        timezone.utc
    ).isoformat(),
}

output_path = (
    PROJECT_DRIVE / "results" / "environment.json"
)

with open(output_path, "w", encoding="utf-8") as f:
    json.dump(environment_info, f, indent=4)

print("Saved:", output_path)
```
